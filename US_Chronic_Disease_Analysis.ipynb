### Introduction to US Chronic Disease Analysis

Chronic disease indicators are standardized public health metrics used to measure the prevalence, incidence, risk factors, healthcare use, and mortality of long-term conditions—also called noncommunicable diseases (NCDs). They help health agencies track population trends, spot disparities across groups or regions, and guide prevention programs.

### Major Chronic Condition Tracked

Surveillance frameworks generally focus on high-burden conditions and their outcomes, such as:
• Cardiovascular disease: High blood pressure, high cholesterol, heart disease mortality, and stroke mortality.
• Diabetes: Diabetes prevalence, gestational diabetes, and related complications.
• Chronic respiratory diseases: Asthma and chronic obstructive pulmonary disease (COPD) prevalence.
• Cancer: Incidence, mortality, and screening rates (such as breast, cervical, and colorectal cancer).
• Other long-term conditions: Chronic kidney disease, arthritis, cognitive health/dementia, disability, and mental health conditions like depression.

The purpose of this analysis is to understand key factors related to chronic disease and present insights through clear and meaningful data visualization.This project uses Python libraries such as Pandas,Numpy,Plotly,Matplotlib,and Seaborn to analyze a US Chronic disease dataset

###  Outlier Analysis

Outlier analysis helps identify unusually high or low values in the datavalue column.

#### Detect outliers using IQR

```python
# Calculate Q1 and Q3
Q1 = df["datavalue"].quantile(0.25)
Q3 = df["datavalue"].quantile(0.75)

# Calculate IQR
IQR = Q3 - Q1

# Calculate limits
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

# Identify outliers
outliers = df[
    (df["datavalue"] < lower_limit) |
    (df["datavalue"] > upper_limit)
]

print("Lower Limit:", lower_limit)
print("Upper Limit:", upper_limit)
print("Total Outliers:", len(outliers))
```

#### Info :-

In this dataset,  first detect outliers and then decide how to handle them.but i should not delete genuine high disease values.

### Handle Outliers

this dataset, keeping the original values and flagging outliers rather than deleting or capping them automatically.

```python
# Handle outliers

df["outlier_status"] = np.where(
    (df["datavalue"] < lower_limit) |
    (df["datavalue"] > upper_limit),
    "Outlier",
    "Normal"
)

print(df["outlier_status"].value_counts())
```

#### Info :-

This creates a new column identifying unusual values without changing original datavalue.
outliers were detected using the Interquartile Range(IQR)method. since this dataset contains chronic disease indicators, extreme  values may represent genuine health conditions.therefore, outliers were flagged for further investigation instead of being automatically removed.
